# Ejemplo de lectura de datos

Este notebook es un ejemplo de cómo utilizar las funciones definidas en `src/tfm_io.py`.

El módulo permite leer los diferentes archivos Excel originales, detectar automáticamente su tipo y transformarlos en DataFrames de pandas con una estructura homogénea.

El objetivo de este notebook es tener un ejemplo del funcionamiento del módulo de lectura de datos.

```python
from pathlib import Path
import sys


PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(
        0,
        str(PROJECT_ROOT),
    )


from src.tfm_io import (
    read_article_sales,
    read_articles,
    read_menu,
    read_departments,
    read_tickets,
    read_tips,
    read_reservations,
    read_data_file,
    load_data_directory,
    summarize_datasets,
)


DATA_DIR = PROJECT_ROOT / 'data'


print(f'Project root: {PROJECT_ROOT}')
print(f'Data directory: {DATA_DIR}')
```

## Lectura individual de archivos

Cada tipo de archivo tiene una función de lectura específica.

Las funciones eliminan la estructura visual innecesaria de los excel, normalizan los nombres de las columnas y devuelven DataFrames más fáciles de utilizar.

```python
article_sales = read_article_sales(
    DATA_DIR / '04-10-05.xls'
)

articles = read_articles(
    DATA_DIR / 'ARTICULOS.xls'
)

total_articles = read_article_sales(
    DATA_DIR / 'TOTAL_ARTICULOS.xls'
)

menu = read_menu(
    DATA_DIR / 'CARTA.xls'
)

departments = read_departments(
    DATA_DIR / 'DEPARTAMENTOS.xls'
)

tickets = read_tickets(
    DATA_DIR / 'LISTA_TICKETS.xls'
)

tips = read_tips(
    DATA_DIR / 'PROPINAS_TICKET.xls'
)

reservations = read_reservations(
    DATA_DIR / 'RESERVAS.xlsx'
)
```

## Visualización de los datasets

Cada función devuelve un DataFrame de pandas.

Mostramos las primeras filas de cada uno de los datasets individuales cargados.

```python
print('ARTICLE SALES')
display(article_sales.head())


print('ARTICLES')
display(articles.head())

print('TOTAL ARTICLES')
display(total_articles.head())

print('MENU')
display(menu.head())


print('DEPARTMENTS')
display(departments.head())


print('TICKETS')
display(tickets.head())


print('TIPS')
display(tips.head())


print('RESERVATIONS')
display(reservations.head())
```

## Detección automática del tipo de archivo

La función `read_data_file()` lee un archivo sin conocer previamente su tipo.

La función analiza el contenido del Excel y devuelve:

1. El nombre del dataset detectado.
2. El DataFrame estandarizado.

```python
dataset_name, data = read_data_file(
    DATA_DIR / 'ARTICULOS.xls'
)


print(f'Detected dataset: {dataset_name}')
print(f'Rows: {len(data)}')
print(f'Columns: {len(data.columns)}')


display(data.head())
```

También podemos comprobar la detección automática de distintos tipos de archivo.

```python
example_files = [
    '04-10-05.xls',
    'TOTAL_ARTICULOS.xls',
    'ARTICULOS.xls',
    'CARTA.xls',
    'DEPARTAMENTOS.xls',
    'LISTA_TICKETS.xls',
    'PROPINAS_TICKET.xls',
    'RESERVAS.xlsx',
]


for file_name in example_files:
    dataset_name, data = read_data_file(
        DATA_DIR / file_name
    )

    print(
        f'{file_name:<25} '
        f'-> {dataset_name:<25} '
        f'rows={len(data)}'
    )
```

## Carga automática de la carpeta de datos

La función `load_data_directory()` busca todos los archivos `.xls` y `.xlsx` de la carpeta de datos, detecta automáticamente su tipo y agrupa los archivos pertenecientes al mismo dataset.

La función devuelve:

1. Un diccionario de DataFrames.
2. Un DataFrame con los errores de carga detectados.

```python
datasets, load_errors = load_data_directory(
    DATA_DIR
)


summary = summarize_datasets(
    datasets
)


display(summary)


if not load_errors.empty:
    print('Files with loading errors:')

    display(load_errors)

else:
    print('All files were loaded correctly.')
```

## Acceso a los datasets

`load_data_directory()` devuelve los datos en un diccionario.

Cada tipo de información puede obtenerse utilizando el nombre del dataset correspondiente.

```python
article_sales_periodic = (
    datasets['article_sales_periodic']
)

article_sales_summary = (
    datasets['article_sales_summary']
)

articles = datasets['articles']

menu = datasets['menu']

departments = datasets['departments']

tickets = datasets['tickets']

tips = datasets['tips']

reservations = datasets['reservations']
```

```python
print('Available datasets:')


for dataset_name in datasets:
    print(f'- {dataset_name}')
```

## Diseño de `tfm_io.py`

El módulo separa las funciones en dos grupos.

### Funciones públicas

Son las funciones que pueden utilizarse directamente desde notebooks y otros módulos del proyecto:

- `read_article_sales()`
- `read_articles()`
- `read_menu()`
- `read_departments()`
- `read_tickets()`
- `read_tips()`
- `read_reservations()`
- `read_data_file()`
- `load_data_directory()`
- `summarize_datasets()`

### Funciones internas

Las funciones cuyo nombre comienza por `_` se utilizan internamente dentro de `tfm_io.py`.

Estas funciones se encargan de tareas como:

- normalizar textos;
- normalizar nombres de columnas;
- interpretar fechas;
- extraer metadatos de los informes;
- detectar el tipo de archivo;
- combinar fechas y horas.

Normalmente no llamamos directamente a estas funciones desde un notebook.

Si utilizamos `load_data_directory()` podemos cargar automáticamente todos los archivos disponibles y obtener los diferentes datasets ya organizados.
